# Make your first API call

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="gpt-5.5",
    input="Explain what an API is in simple English"
)

print(response.output_text)
```

# A more useful example

Imagine you are building an AI application that summarizes customer reviews.

```python
from openai import OpenAI

client = OpenAI()

review = """
The laptop looks great and the screen is excellent.
However, the battery life is disappointing.
"""

response = client.responses.create(
    model="gpt-5.5",
    instructions="You are a helpful customer review analyst.",
    input=f"""
Analyze this customer review.

Review:
{review}

Give me:
1. Sentiment
2. Main positive point
3. Main negative point
""",
)

print(response.output_text)
```
