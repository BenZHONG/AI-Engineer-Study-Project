# System Prompt

A **System Prompt** is a set of instructions that tells an AI model **how it should behave, what role it should play, and what rules it should follow**.

Think of it as the AI's **operating instructions**.

## 1. A simple example

Imagine you are building a customer-support chatbot.

You might give it this system prompt:

```text
You are a customer support assistant for an online store.

Your responsibilities:
- Help customers with orders and returns.
- Be polite and professional.
- Give concise answers.
- If you don't know the answer, say so.
- Never invent order information.
```

Then a user asks:

```text
Where is my order?
```

The system prompt tells the AI **how to approach the question**.

---

## 2. System Prompt vs User Prompt

This is one of the most important concepts in prompt engineering.

### System Prompt

Defines the AI's **role, behavior, rules, and constraints**.

```text
You are a Python programming teacher.
Explain concepts step-by-step.
Assume the student is a beginner.
Use simple examples.
```

### User Prompt

Contains the **specific request** from the user.

```text
What is a Python dictionary?
```

The model combines the two:

```text
System:
You are a Python programming teacher.
Explain concepts step-by-step.
Assume the student is a beginner.

User:
What is a Python dictionary?
```

The expected response would therefore be a **beginner-friendly explanation**, rather than an advanced academic explanation.

---

## 3. A useful mental model

You can think about it like this:

![](assets/01.%20Prompt%20Engineering/file-20260917165123957.png)

The **system prompt establishes the context**, while the **user prompt asks for the actual task**.

---

## 4. What can a System Prompt contain?

A good system prompt often has several components.

### Role

Tell the model who it is.

```text
You are an experienced Python teacher.
```

### Goal

Tell it what it should accomplish.

```text
Your goal is to help students understand Python programming.
```

### Behavior

Tell it how to respond.

```text
Explain difficult concepts using simple language.
```

### Constraints

Tell it what it should or shouldn't do.

```text
Do not assume the student has advanced programming knowledge.
```

### Output format

Tell it how to structure its response.

```text
When explaining code:
1. Explain the concept.
2. Show an example.
3. Explain the example line by line.
```

Putting these together:

```text
You are an experienced Python teacher.

Your goal is to help beginners learn Python.

Rules:
- Use simple English.
- Explain concepts step-by-step.
- Give practical examples.
- Do not assume advanced programming knowledge.
- When showing code, explain important lines.

When answering a question:
1. Explain the concept.
2. Give a simple example.
3. Explain the example.
4. Give a small practice exercise.
```

That's a much more useful system prompt than simply:

```text
You are a Python teacher.
```

---

## 5. Why is the System Prompt important?

Because the same user question can produce very different answers depending on the system prompt.

For example:

**System Prompt A**

```text
You are a senior software engineer.
Use technical terminology.
Assume the reader has 5 years of programming experience.
```

User:

```text
What is an API?
```

The answer might discuss:

- HTTP
    
- REST
    
- endpoints
    
- authentication
    
- serialization
    
- status codes

---

**System Prompt B**

```text
You are a teacher for children.
Explain everything using simple language and everyday examples.
```

Same user:

```text
What is an API?
```

The answer might be:

> An API is like a waiter in a restaurant. You tell the waiter what you want, the waiter takes your request to the kitchen, and then brings the result back to you.

Same model. Same user prompt. **Different system instructions → different behavior.**

---

## 6. System Prompt in an AI application

When you build an AI application using an API, you commonly have something like:

```python
messages = [
    {
        "role": "system",
        "content": """
        You are a helpful Python teacher.
        Explain concepts step-by-step.
        Use simple examples.
        """
    },
    {
        "role": "user",
        "content": "What is a list in Python?"
    }
]
```

Conceptually:

```text
role = system
        ↓
"How should you behave?"

role = user
        ↓
"What should you do?"
```

Modern APIs may represent these instructions differently, but the **concept of separating higher-level instructions from the user's task remains important**.

## 7. System Prompt and Prompt Engineering

When you study **Prompt Engineering**, System Prompts are one of the fundamental techniques.

You are essentially controlling four things:

![](assets/01.%20Prompt%20Engineering/file-20260917165203935.png)

For example:

```text
ROLE
You are an AI Engineer with 5 years of experience.

TASK
Help me debug Python code.

CONSTRAINTS
- Explain the cause of the error.
- Provide corrected code.
- Explain the correction.
- Don't rewrite unrelated parts.

OUTPUT
Use:
1. Problem
2. Cause
3. Solution
4. Explanation
```

This is much closer to **professional prompt engineering** than simply asking:

```text
Fix my code.
```

---

## 8. One important distinction

A System Prompt is **not magic** and it isn't an absolute guarantee that the model will follow every instruction.

A model receives a hierarchy of instructions and context. In a typical application, you can think conceptually about:

```text
Higher-priority instructions
          ↓
      System
          ↓
     Developer
          ↓
        User
          ↓
    Model response
```

The exact message hierarchy depends on the API/platform, but the important prompt-engineering lesson is:

> **Instructions have different levels of authority.**

This becomes especially important when you build **AI agents, RAG systems, tool-calling applications, and MCP applications**.

## 9. A real-world example: RAG application

Since you're learning **RAG and AI Engineering**, imagine you're building a company-document chatbot.

Your system prompt could be:

```text
You are a company knowledge assistant.

Answer questions using the information provided
in the retrieved documents.

Rules:
- Only use information supported by the retrieved documents.
- Do not invent facts.
- If the documents don't contain the answer, say:
  "I don't have enough information to answer this question."
- Keep answers concise.
- Cite the relevant document when possible.
```

Then your application supplies retrieved documents:

```text
System Prompt
      +
User Question
      +
Retrieved Documents
      ↓
    LLM
      ↓
   Answer
```

For example:

```text
User:
How many days of annual leave do employees receive?
```

The RAG system retrieves:

```text
Employee Handbook:
Full-time employees receive 20 days of annual leave per year.
```

The model uses the **system prompt + retrieved context + user question** to produce the answer.

## 10. A good System Prompt template

When you're building AI applications, this template is useful:

```text
# Role
You are [ROLE].

# Objective
Your goal is to [OBJECTIVE].

# Responsibilities
You should:
- [RESPONSIBILITY 1]
- [RESPONSIBILITY 2]
- [RESPONSIBILITY 3]

# Rules
- [RULE 1]
- [RULE 2]
- [RULE 3]

# Constraints
- [CONSTRAINT 1]
- [CONSTRAINT 2]

# Output Format
Return your response in the following format:
[FORMAT]
```

For example:

```text
# Role
You are a senior Python programming teacher.

# Objective
Help beginners learn Python programming.

# Rules
- Use simple English.
- Explain concepts step-by-step.
- Provide practical examples.
- Do not assume advanced knowledge.

# Output Format
For programming questions, provide:
1. Explanation
2. Code example
3. Line-by-line explanation
4. Practice exercise
```

If you're learning Prompt Engineering, remember this distinction:

> **User Prompt = What the user wants the AI to do.**

And when you start building production AI systems:

```text
System Prompt
      ↓
defines behavior

User Prompt
      ↓
defines request

Context / RAG
      ↓
provides knowledge

Tools
      ↓
provide capabilities

LLM
      ↓
generates response
```

# User Prompt?

A **User Prompt** is the instruction, question, request, or information that a user sends to an AI model.

In simple terms:

> **User Prompt = What the user asks the AI to do.**

For example:

```text
What is Python?
```

This is a **User Prompt**.

The AI reads the prompt and generates a response based on it.

---

## 1. Simple example

Imagine you are talking to ChatGPT:

**User:**

```text
Explain what AWS is.
```

The text:

```text
Explain what AWS is.
```

is the **User Prompt**.

The AI might respond:

```text
AWS (Amazon Web Services) is a cloud computing platform...
```

So the basic interaction is:

```text
User Prompt
     ↓
   AI Model
     ↓
AI Response
```

---

## 2. User Prompt vs System Prompt

This is an important distinction in prompt engineering.

### System Prompt

The **System Prompt** defines the AI's high-level behavior, rules, role, and constraints.

For example:

```text
You are an experienced Python teacher.
Explain technical concepts in simple language.
Use examples when appropriate.
```

### User Prompt

The **User Prompt** contains the user's actual request:

```text
Explain Python decorators.
```

Together:

```text
System Prompt
    ↓
"You are an experienced Python teacher..."

User Prompt
    ↓
"Explain Python decorators."

        ↓

     AI Model

        ↓

   Final Response
```

You can think of it like:

|Component|Purpose|
|---|---|
|**System Prompt**|Defines how the AI should behave|
|**User Prompt**|Tells the AI what the user wants|
|**Assistant Response**|AI's answer|

---

## 3. A User Prompt can be very simple

For example:

```text
What is Python?
```

or:

```text
What is RAG?
```

or:

```text
Translate this sentence into Chinese.
```

These are all User Prompts.

---

## 4. A User Prompt can also be very detailed

For example:

```text
You are a Python teacher.

Explain Python decorators to me.

My Python level is beginner.

Please:
1. Explain the concept in simple English.
2. Give me a real-world analogy.
3. Show me a simple code example.
4. Explain the code line by line.
5. Give me a small exercise at the end.
```

This is still **one User Prompt**, but it contains several components.

We can break it down:

```text
Role
 ↓
"You are a Python teacher."

Context
 ↓
"My Python level is beginner."

Task
 ↓
"Explain Python decorators."

Instructions
 ↓
"Explain in simple English."
"Give an analogy."
"Show code."
"Explain line by line."

Output requirements
 ↓
"Give me an exercise."
```

This is where **prompt engineering** becomes useful.

---

## 5. The main components of a good User Prompt

A useful way to design prompts is:

```text
User Prompt
│
├── Task
├── Context
├── Instructions
├── Constraints
├── Examples
└── Output Format
```

Let's look at each one.

### Task

What do you want the AI to do?

```text
Explain RAG.
```

### Context

Give the AI information it needs.

```text
I am a beginner in AI and already know Python.
```

### Instructions

Tell the AI how to perform the task.

```text
Explain the concept step by step.
```

### Constraints

Tell the AI what limitations to follow.

```text
Use simple English and keep the explanation under 500 words.
```

### Examples

Show the AI what you want.

```text
Example:

Input: "What is Python?"
Style: Simple explanation with an analogy.
```

### Output Format

Tell the AI how you want the answer structured.

```text
Return the answer using:

1. Definition
2. Example
3. Code
4. Exercise
```

---

## 6. Example: Weak vs Strong User Prompt

### Weak prompt

```text
Teach me AWS.
```

This is valid, but it doesn't give the AI much information.

The AI doesn't know:

- Your current level
    
- What AWS topics you need
    
- How much time you have
    
- Whether you want theory or practice
    
- What your goal is

---

### Stronger prompt

```text
I am a beginner in AWS.

I already know basic Python and networking concepts.

My goal is to prepare for an entry-level Cloud Engineer position.

Create a 4-week AWS study plan.

I can study 10 hours per week.

For each week, provide:
1. Topics to learn
2. Learning resources
3. Hands-on exercises
4. A small project
5. Interview questions
```

This is a much more useful **User Prompt**.

The AI now has:

```text
Context
    ↓
Beginner + Python/networking knowledge

Goal
    ↓
Cloud Engineer preparation

Task
    ↓
Create 4-week study plan

Constraint
    ↓
10 hours/week

Output format
    ↓
Topics + resources + exercises + projects + interview questions
```

---

## 7. User Prompt in an AI application

This becomes even more important when you build an AI application.

For example, imagine you build a customer-support chatbot.

The user types:

```text
My AWS EC2 instance is not responding.
```

That is the **User Prompt**.

Your application might combine it with other information:

```text
System Prompt:
You are an AWS technical support assistant.

User Prompt:
My AWS EC2 instance is not responding.

Additional Context:
EC2 instance ID: i-123456
Region: ap-southeast-2
Status: Running
```

The model can then produce a much more useful answer.

This is a fundamental idea in AI application development:

```text
System Instructions
        +
User Prompt
        +
Context/Data
        ↓
      LLM
        ↓
     Response
```

This is particularly important for **RAG systems**.

---

## 8. User Prompt in RAG

Suppose you build a company-document chatbot.

The user asks:

```text
How many days of annual leave do employees receive?
```

The user prompt itself may be:

```text
How many days of annual leave do employees receive?
```

But your RAG application might retrieve relevant information from company documents:

```text
User Prompt
     ↓
"How many days of annual leave..."
     ↓
Vector Search
     ↓
Relevant company documents
     ↓
LLM
     ↓
Answer
```

The model might effectively receive:

```text
System:
You are a company HR assistant.
Answer using the provided company documents.

Context:
[Relevant HR policy]

User:
How many days of annual leave do employees receive?
```

So the **User Prompt** is still the user's question.

The retrieved documents are **context**, not necessarily part of the original user prompt.

---

## 9. User Prompt in an API

When you use an LLM API, you will often see something conceptually like:

```python
response = client.responses.create(
    model="...",
    instructions="You are a helpful Python teacher.",
    input="Explain Python decorators."
)
```

Here:

```text
instructions
    ↓
System-level instructions

input
    ↓
User Prompt
```

So you can think of:

```python
input="Explain Python decorators."
```

as the user's prompt.

---

## 10. An important prompt-engineering principle

A User Prompt is **not just a question**.

It can contain:

```text
Question
Instruction
Context
Examples
Constraints
Data
Output format
```

For example:

```text
I am a beginner Python developer.

I want to learn RAG.

Explain RAG using a simple example.

Compare:
- Traditional LLM
- RAG

Use a table.

Then provide a small Python example using LangChain.
```

This entire piece of text is a **User Prompt**.

---

## 11. A useful mental model

As you learn Prompt Engineering, remember this:

```text
                PROMPT
                   │
        ┌──────────┴──────────┐
        │                     │
   System Prompt         User Prompt
        │                     │
  "How should you       "What should
   behave?"              you do?"
        │                     │
        └──────────┬──────────┘
                   ↓
                  LLM
                   ↓
               Response
```

The simplest way to remember it is:

> **System Prompt = How the AI should behave.**
> **User Prompt = What the user wants the AI to do.**

And in practical prompt engineering, the quality of the User Prompt often improves when you clearly specify **the task, context, constraints, and desired output format**.

# Few-shot Prompting

**Few-shot prompting** is a prompt engineering technique where you give an AI model **a few examples of the task you want it to perform**, and then ask it to perform the same task on new input.

In simple terms:

> **“Here are a few examples of what I want. Now do the same thing for this new case.”**

---

## 1. Why is it called "Few-shot"?

The word **“shot”** means an example provided to the model.

There are three common approaches:

| Technique     | Examples provided | Meaning                            |
| ------------- | ----------------: | ---------------------------------- |
| **Zero-shot** |                 0 | Give instructions, but no examples |
| **One-shot**  |                 1 | Give one example                   |
| **Few-shot**  |             A few | Give several examples              |

For example:

### Zero-shot

```text
Classify the sentiment as Positive or Negative.

Text: "The product is excellent!"
```

The model has to understand the task from the instruction alone.

---

### One-shot

```text
Classify the sentiment as Positive or Negative.

Example:
Text: "I love this product."
Sentiment: Positive

Now classify:
Text: "This product is terrible."
Sentiment:
```

You give **one example**.

---

### Few-shot

Now we provide several examples:

```text
Classify the sentiment as Positive or Negative.

Example 1:
Text: "I love this product."
Sentiment: Positive

Example 2:
Text: "This is the worst service I've ever received."
Sentiment: Negative

Example 3:
Text: "The product works perfectly."
Sentiment: Positive

Example 4:
Text: "I am very disappointed."
Sentiment: Negative

Now classify:
Text: "The customer service was fantastic."
Sentiment:
```

The model can infer the pattern from the examples:

```text
Positive → good experience
Negative → bad experience
```

So it should produce:

```text
Positive
```

---

## 2. Few-shot is more than just giving examples

The important idea is that **the examples demonstrate the behavior you want**.

For example, suppose you want an AI to convert informal English into professional English.

Instead of only saying:

```text
Rewrite the following sentence professionally.
```

you can provide examples:

```text
Convert informal English into professional English.

Example 1:
Input: "Can you send me the file?"
Output: "Could you please send me the file?"

Example 2:
Input: "I can't come to the meeting."
Output: "Unfortunately, I won't be able to attend the meeting."

Example 3:
Input: "Give me more information."
Output: "Could you please provide me with some additional information?"

Now rewrite:
Input: "I need this report tomorrow."
Output:
```

The model learns the **style and pattern** from the examples.

Possible output:

```text
Could you please provide the report by tomorrow?
```

---

## 3. A useful way to think about Few-shot

Think of it like **teaching a new employee**.

Imagine you tell a new employee:

> "When customers ask questions, reply professionally."

That's similar to **zero-shot prompting**.

But if you show them:

```text
Customer: "Where is my order?"
Employee: "I'd be happy to check the status of your order for you."

Customer: "Can I change my address?"
Employee: "Certainly. I can help you update your delivery address."
```

Then give them a new customer question:

```text
Customer: "Can I cancel my order?"
```

They can imitate the pattern.

That's essentially **few-shot prompting**.

---

## 4. Few-shot for structured output

Few-shot prompting is particularly useful when you want the model to follow a **specific format**.

For example:

```text
Extract the person's name and age.

Example 1:
Input: "John is 25 years old."
Output:
Name: John
Age: 25

Example 2:
Input: "Sarah, who is 31, lives in Sydney."
Output:
Name: Sarah
Age: 31

Now process:
Input: "David is 42 years old."
Output:
```

The model can follow the demonstrated format:

```text
Name: David
Age: 42
```

This is useful for:

* JSON generation
* Data extraction
* Classification
* Text transformation
* Sentiment analysis
* Information extraction
* Code generation
* Customer-service responses

---

## 5. Few-shot with JSON

Suppose you're building an AI application and want consistent JSON.

You could write:

```text
Extract the product information and return JSON.

Example 1:

Input:
"Apple iPhone 15 costs $999."

Output:
{
  "product": "Apple iPhone 15",
  "price": 999
}

Example 2:

Input:
"Samsung Galaxy S24 costs $899."

Output:
{
  "product": "Samsung Galaxy S24",
  "price": 899
}

Now process:

Input:
"Google Pixel 9 costs $799."

Output:
```

The model learns the desired structure:

```json
{
  "product": "Google Pixel 9",
  "price": 799
}
```

---

## 6. Few-shot for classification

This is another very common application.

Imagine you're building a support-ticket system.

You want to classify tickets into:

* `Billing`
* `Technical`
* `Account`

You can provide examples:

```text
Classify the support request.

Example 1:
"Why was I charged twice?"
Category: Billing

Example 2:
"My application keeps crashing."
Category: Technical

Example 3:
"I forgot my password."
Category: Account

Now classify:

"I cannot log into my account."
Category:
```

The model should infer:

```text
Account
```

Notice that the examples teach the model **what the categories mean** without requiring a long definition for every category.

---

## 7. Few-shot vs Fine-tuning

This is an important distinction.

### Few-shot prompting

You put examples **inside the prompt**:

```text
Example 1 → Answer 1
Example 2 → Answer 2
Example 3 → Answer 3

New input → ?
```

You don't change the model itself.

### Fine-tuning

You train a model using a dataset of examples so that the model's behavior is changed for that task.

A simple comparison:

```text
Few-shot:

Prompt
   ↓
Examples
   ↓
LLM
   ↓
Answer
```

Whereas:

```text
Fine-tuning:

Training examples
       ↓
    Training
       ↓
Fine-tuned model
       ↓
      Prompt
       ↓
     Answer
```

So **few-shot is a prompting technique**, while **fine-tuning is a model-training technique**.

---

## 8. How many examples should you provide?

There isn't a fixed number.

"Few" might mean:

```text
2 examples
3 examples
5 examples
10 examples
```

depending on the task and available context.

The important thing is **quality and relevance**, not simply the number of examples.

For example, if you're teaching sentiment classification, these examples are useful:

```text
"I love it." → Positive
"I hate it." → Negative
"Fantastic service." → Positive
"Terrible service." → Negative
```

But these examples may not be very useful:

```text
"I love it." → Positive
"London is in England." → Geography
"Python is a programming language." → Programming
"Pizza is delicious." → Positive
```

The examples should be **representative of the task**.

---

## 9. A good Few-shot prompt structure

A useful template is:

```text
[Instruction]

Example 1:
Input: ...
Output: ...

Example 2:
Input: ...
Output: ...

Example 3:
Input: ...
Output: ...

Now process:
Input: ...
Output:
```

For example:

```text
Translate English into simple business English.

Example 1:
Input: "Send me the document."
Output: "Could you please send me the document?"

Example 2:
Input: "I need more information."
Output: "Could you please provide me with more information?"

Example 3:
Input: "Tell me when you are available."
Output: "Could you please let me know when you are available?"

Now rewrite:
Input: "Give me the report today."
Output:
```

---

## 10. The key idea

The most important concept to remember is:

> **Few-shot prompting teaches the model through examples rather than only through instructions.**

You can remember the three levels like this:

```text
ZERO-SHOT
Instruction only
       ↓
     LLM
       ↓
    Answer


ONE-SHOT
Instruction
+ 1 example
       ↓
     LLM
       ↓
    Answer


FEW-SHOT
Instruction
+ several examples
       ↓
     LLM
       ↓
    Answer
```

### In one sentence:

**Few-shot prompting = Give the LLM a few examples of input → output, so it can infer the pattern and apply it to new input.**

Since you're learning prompt engineering, the next useful concept is **Few-shot vs Zero-shot vs Chain-of-Thought**, because these three techniques are often discussed together.


# Role Prompting

**Role prompting** is a prompt engineering technique where you tell an AI model to **take on a specific role, profession, or perspective** before asking it to perform a task.

In simple terms:

> **Role prompting = Tell the AI "who you want it to be" before telling it "what you want it to do."**

---

## 1. Basic Example

Without role prompting:

```text
Explain AWS VPC.
```

With role prompting:

```text
You are an AWS Cloud Engineer.
Explain AWS VPC to a beginner.
Use simple English and provide practical examples.
```

The second prompt gives the AI additional context about **how it should approach the answer**.

---

## 2. Common Structure

A role prompt often looks like this:

```text
You are a [ROLE].

Your task is to [TASK].

Please [REQUIREMENTS].
```

For example:

```text
You are a Python teacher.

Explain Python decorators to a beginner.

Use simple English and provide three practical examples.
```

Here:

- **Role:** Python teacher
    
- **Task:** Explain decorators
    
- **Audience:** Beginner
    
- **Style:** Simple English
    
- **Requirement:** Three examples
    

---

## 3. Why Use Role Prompting?

Role prompting can help guide the model toward a particular **perspective, level of expertise, tone, and style**.

### Example: Different roles

The same question can produce different types of answers depending on the role.

**Role: Teacher**

```text
You are a Python teacher.
Explain what a Python list is to a beginner.
```

Focus:

- Simple explanation
    
- Teaching
    
- Examples
    
- Beginner-friendly language
    

**Role: Senior Python Developer**

```text
You are a senior Python developer.
Explain Python lists and discuss when they should be used.
```

Focus:

- Practical programming
    
- Best practices
    
- Performance considerations
    
- Real-world usage
    

**Role: Interviewer**

```text
You are a technical interviewer.
Ask me five Python interview questions about lists.
Ask one question at a time.
```

Focus:

- Interview questions
    
- Evaluation
    
- Interactive conversation
    

---

## 4. Role Prompting + Task

A useful pattern is:

```text
You are a [ROLE].

Your goal is to [GOAL].

Please [TASK].

Constraints:
- ...
- ...
- ...
```

For example:

```text
You are an experienced AI Engineer.

Your goal is to help me learn RAG.

Explain what Retrieval-Augmented Generation is.

I am a beginner, so:
- Use simple English.
- Explain the concepts step by step.
- Give a practical example.
- Do not assume prior knowledge.
```

This is much more specific than simply asking:

```text
What is RAG?
```

---

## 5. Role Prompting Does NOT Give the AI Real-World Experience

This is an important point.

If you write:

```text
You are a doctor with 20 years of experience.
```

it does **not** mean the AI actually has 20 years of experience.

The role is a **behavioral instruction**. It tells the model what perspective and style to adopt.

For example:

```text
You are a senior software engineer.
Review the following Python code and explain potential problems.
```

This encourages the model to approach the code like a senior engineer would, but you should still verify important technical information.

---

## 6. Role Prompting vs. System Prompt

These concepts are related but different.

### System prompt

A **system prompt** defines high-level instructions for the AI.

```text
You are a helpful programming tutor.
Always explain technical concepts clearly.
```

### Role prompting

Role prompting is specifically about assigning a **role/persona/perspective**.

```text
You are a senior Python developer.
```

So role prompting can be **part of a system prompt or another prompt**.

---

## 7. Role Prompting in Your AI Projects

Since you're learning **AI Engineering and Prompt Engineering**, you might use roles like:

```text
You are a Prompt Engineering teacher.
```

or:

```text
You are an AI Engineer specializing in RAG systems.
```

or:

```text
You are a Python debugging assistant.
```

For example:

## 8. A Key Idea to Remember

```text
Role + Goal + Task + Constraints
```

```text
You are a Python teacher.
Help me learn Python decorators.
Explain them step by step using simple English
and practical examples.
```

# Structured Output

**Structured output** means asking an AI to return information in a **specific format or structure**, instead of letting it answer freely.

For example, instead of asking:

> "Tell me about this product."

You can ask the AI to return:

```json
{
  "name": "iPhone 17",
  "price": 999,
  "category": "Smartphone",
  "available": true
}
```

This is useful when your application needs to **read and use the AI's answer automatically**.

---

## 1. What problem does Structured Output solve?

Normally, an LLM can return answers in many different ways.

For example:

```text
The customer is John. He is 32 years old and lives in Sydney.
```

Another time, it might return:

```text
Customer: John
Age: 32
Location: Sydney
```

Both answers are understandable to humans.

But for a Python program, this can be difficult to process reliably.

Your program would prefer something predictable:

```json
{
  "name": "John",
  "age": 32,
  "location": "Sydney"
}
```

Now your Python code can easily do:

```python
data["name"]
data["age"]
data["location"]
```

---

## 2. Structured Output = Defined Structure

The basic idea is:

**You tell the AI what fields you want and what type of data each field should contain.**

For example:

```text
name: string
age: integer
city: string
```

The AI should produce:

```json
{
  "name": "John",
  "age": 32,
  "city": "Sydney"
}
```

Think of it like a **form**.

Instead of saying:

> "Give me some customer information."

You give the AI a form:

|Field|Type|
|---|---|
|name|string|
|age|integer|
|city|string|
|email|string|

The AI fills in the form.

---

## 3. Structured Output vs Normal Output

### Normal output

Prompt:

```text
Tell me the name, age and city of this customer:

John is 32 years old and lives in Sydney.
```

Possible answer:

```text
The customer is John. He is 32 and lives in Sydney.
```

This is good for a human.

### Structured output

Prompt:

```text
Return the customer information in JSON.

{
  "name": string,
  "age": integer,
  "city": string
}
```

Answer:

```json
{
  "name": "John",
  "age": 32,
  "city": "Sydney"
}
```

This is much easier for software to process.

---

## 4. Why is Structured Output important?

Structured output is especially useful when building **AI applications**.

For example:

### Customer support

An AI reads:

> "My order #12345 arrived damaged."

You might want:

```json
{
  "order_id": "12345",
  "problem": "damaged",
  "priority": "high"
}
```

Your application can then automatically create a support ticket.

---

### Sentiment analysis

Input:

> "The product is great, but delivery was very slow."

Output:

```json
{
  "sentiment": "mixed",
  "positive": true,
  "negative": true
}
```

---

### Extracting information from documents

Input:

```text
John Smith
Email: john@example.com
Age: 35
```

Output:

```json
{
  "name": "John Smith",
  "email": "john@example.com",
  "age": 35
}
```

---

## 5. Bad Prompt Example

Here is a weak prompt:

```text
Read this customer information and give me the customer's name,
age, email and city.

John Smith is 35 years old. His email is john@example.com.
He lives in Sydney.
```

The AI might answer:

```text
The customer's name is John Smith. He is 35 years old,
his email is john@example.com, and he lives in Sydney.
```

**The format is not clearly defined.**

Your program might receive different formats from different requests.

Expected output:

```json
{
  "name": "John Smith",
  "age": 35,
  "email": "john@example.com",
  "city": "Sydney"
}
```

---

## 7. Why is the Improved Prompt Better?

There are several improvements.

### It defines the required fields

The prompt says:

```text
"name"
"age"
"email"
"city"
```

The AI knows exactly what information to return.

---

### It defines data types

For example:

```text
age: integer
```

So we want:

```json
"age": 35
```

rather than:

```json
"age": "He is thirty-five years old"
```

This makes the output easier for a program to use.

---

### It defines the format

The prompt says:

```text
Return ONLY valid JSON.
```

Therefore, we don't want:

```text
Sure! Here is the information:

{
    "name": "John Smith",
    ...
}
```

We want just the structured data:

```json
{
    "name": "John Smith",
    "age": 35,
    "email": "john@example.com",
    "city": "Sydney"
}
```

---

### It makes the output predictable

Suppose you process 1,000 customers.

You want every response to have the same structure:

```json
{
  "name": "...",
  "age": 0,
  "email": "...",
  "city": "..."
}
```

Predictable output makes it much easier for your application to process thousands of responses.

**JSON is one of the most common structured output formats.**

# JSON Output

As a Prompt Engineering teacher, let's learn **JSON output** step by step.

## 1. What is JSON?

**JSON** stands for **JavaScript Object Notation**.

It is a simple format for storing and exchanging data.

For example:

```json
{
  "name": "John",
  "age": 30,
  "city": "Sydney"
}
```

This data has three pieces of information:

- `name` → `"John"`
    
- `age` → `30`
    
- `city` → `"Sydney"`

The important thing is that the data has a **clear structure**.

---

## 2. What is JSON output?

**JSON output means asking an AI model to return its answer in JSON format.**

Instead of asking:

> Tell me about this product.

The AI might answer:

> This is a laptop. It costs $1,200 and is suitable for students.

This is easy for a human to read, but it can be difficult for a computer program to process.

With JSON output, we can ask the AI to return:

```json
{
  "product": "Laptop",
  "price": 1200,
  "suitable_for": "students"
}
```

Now a Python program, JavaScript program, or API can easily use the information.

---

## 3. Why is JSON output useful?

JSON output is especially useful when **another program needs to use the AI's answer**.

For example, imagine you build an AI system that analyzes customer reviews.

A customer writes:

> The product arrived quickly, but the battery life is terrible.

You could ask the AI to return:

```json
{
  "sentiment": "negative",
  "topic": "battery",
  "delivery": "positive"
}
```

Your application can then use these fields automatically.

For example:

```python
result["sentiment"]
```

would give:

```text
negative
```

---

## 4. Basic JSON structure

There are a few important rules.

### Object

JSON objects use `{ }`.

```json
{
  "name": "Alice",
  "age": 25
}
```

### Key-value pairs

Information is stored as:

```text
"key": value
```

For example:

```json
"name": "Alice"
```

### Strings

Strings use double quotes:

```json
"name": "Alice"
```

### Numbers

Numbers do not need quotes:

```json
"age": 25
```

### Arrays

JSON can also contain lists:

```json
{
  "name": "Alice",
  "skills": [
    "Python",
    "SQL",
    "AWS"
  ]
}
```

---

## 5. Bad prompt example

Here is a weak prompt:

> Analyze this customer review and give me the sentiment, product, and important information.

The AI might respond:

> The review is negative. The customer is talking about a laptop. The main problem is the battery. They also mentioned that delivery was fast.

This is useful for a human, but it has a problem:

**The format is unpredictable.**

One response might say:

```text
The sentiment is negative...
```

Another might say:

```text
This is a negative review...
```

Another might use bullet points.

That makes it harder for your application to process the response automatically.

---

## 6. Improved prompt example

We can explicitly request JSON:

```text
Analyze the customer review below.

Return the result as JSON.

Use exactly these fields:
- sentiment
- product
- main_issue
- delivery

Rules:
1. Return valid JSON only.
2. Do not include Markdown.
3. Do not include explanations outside the JSON.
4. Use "positive", "negative", or "neutral" for sentiment.
5. If information is not available, use null.

Customer review:
"The laptop arrived very quickly, but the battery life is terrible."
```

A good response would be:

```json
{
  "sentiment": "negative",
  "product": "laptop",
  "main_issue": "battery life",
  "delivery": "positive"
}
```

---

## 7. Why is the improved prompt better?

There are several improvements.

### It clearly specifies the output format

We say:

> Return the result as JSON.

The model knows that we don't want normal prose.

---

### It specifies the fields

We explicitly define:

```text
sentiment
product
main_issue
delivery
```

This creates a predictable structure.

---

### It specifies allowed values

We say:

> Use "positive", "negative", or "neutral" for sentiment.

This reduces variations such as:

```text
"bad"
"unhappy"
"not good"
"negative sentiment"
```

Instead, the application can expect:

```text
"positive"
"negative"
"neutral"
```

---

### It handles missing information

We say:

> If information is not available, use null.

For example:

```json
{
  "sentiment": "positive",
  "product": "laptop",
  "main_issue": null,
  "delivery": "positive"
}
```

This is much better than having the AI invent information.

---

### It prevents extra text

We say:

> Return valid JSON only.

and:

> Do not include explanations outside the JSON.

Without this instruction, the model might produce:

```text
Sure! Here is the JSON:

{
  "sentiment": "negative"
}
```

That extra text can cause problems if your program expects JSON directly.

---

These two ideas are closely related, but they are not exactly the same.

**JSON output** means:

> "Please give me the answer in JSON format."

For example:

```json
{
  "name": "John",
  "age": 30
}
```

It means you define the **exact structure and rules** that the output must follow.

So you can think of it like this:

```text
JSON Output
     ↓
A structured data format

Structured Output
     ↓
A defined and often enforced schema
```

# Prompt Templates

A **prompt template** is a reusable prompt with **placeholders** for information that can change.

Instead of writing a completely new prompt every time, you create one template and insert different values into it.

---

## 1. The basic idea

Imagine you want an AI to translate text.

Without a template, you might write:

> Translate "Hello, how are you?" into Chinese.

Next time:

> Translate "I like Python." into Chinese.

And again:

> Translate "Where is the train station?" into Chinese.

This works, but you are repeating the same instructions.

A **prompt template** solves this problem:

```text
Translate the following text into {language}:

{text}
```

Here:

- `{language}` = a variable
    
- `{text}` = a variable

You can reuse the same template:

```text
language = "Chinese"
text = "Hello, how are you?"
```

The final prompt becomes:

```text
Translate the following text into Chinese:

Hello, how are you?
```

---

## 2. Why do we use Prompt Templates?

Prompt templates are useful when you need to perform the **same task many times**.

For example:

```text
Summarise this {document_type} in {number} sentences:

{document}
```

You can use it for:

- articles
    
- emails
    
- reports
    
- books
    
- customer feedback
    
- meeting notes

Only the variables change.

---

## 3. Prompt Template structure

A prompt template usually contains two parts:

### Fixed instructions

These stay the same.

```text
Summarise the following text in 3 sentences.
Use simple English.
```

### Variables

These change for each request.

```text
{text}
```

Together:

```text
Summarise the following text in 3 sentences.
Use simple English.

{text}
```

You can think of it like a **Python function**.

For example:

```python
def greet(name):
    return f"Hello, {name}!"
```

You don't create a new function for every person.

You reuse the same function:

```python
greet("John")
greet("Mary")
greet("David")
```

A prompt template works in a similar way:

```text
Translate {text} into {language}.
```

Then:

```text
text = "I like programming"
language = "Chinese"
```

---

## 4. Bad Prompt Example

Suppose you are building a customer-support application.

You want AI to answer customer questions.

A bad approach might be:

```text
Answer this question:

How can I reset my password?
```

Then another request:

```text
Answer this question:

How can I change my email address?
```

And another:

```text
Answer this question:

How can I cancel my account?
```

### What's wrong?

The prompt is not reusable.

You are manually rewriting the prompt every time.

It also doesn't clearly tell the AI:

- what role it should have
    
- how detailed the answer should be
    
- what style to use
    
- where the changing information goes

---

## 5. Improved Prompt Template

A better template would be:

```text
You are a helpful customer support assistant.

Answer the customer's question using simple and clear English.

Customer question:
{question}

Requirements:
- Be polite.
- Give step-by-step instructions when appropriate.
- Do not invent information.
```

Now `{question}` is the variable.

For example:

```text
question = "How can I reset my password?"
```

The application can generate:

```text
You are a helpful customer support assistant.

Answer the customer's question using simple and clear English.

Customer question:
How can I reset my password?

Requirements:
- Be polite.
- Give step-by-step instructions when appropriate.
- Do not invent information.
```

For another customer:

```text
question = "How can I change my email address?"
```

The **template stays the same**.

Only `{question}` changes.

---

## 6. Multiple Variables

A template can have more than one variable.

For example:

```text
You are a {role}.

Explain {topic} to a {audience}.

Use {language}.
Keep the explanation under {word_count} words.
```

You could provide:

```text
role = "Python teacher"
topic = "list comprehensions"
audience = "beginner"
language = "English"
word_count = "300"
```

The AI then receives a complete prompt based on those values.

## 7. Prompt Templates in Python

```python
topic = "Python lists"
level = "beginner"

prompt = f"""
You are a Python teacher.

Explain {topic} to a {level} student.
Use simple English.
Give two examples.
"""
```

The resulting prompt is:

```text
You are a Python teacher.

Explain Python lists to a beginner student.
Use simple English.
Give two examples.
```

```text
Template + Variables → Final Prompt
```

The improved prompt is better for several reasons.

Then use it thousands of times.

### Consistent

## 9. A simple way to remember it

Think of a **prompt template as a form**.

For example:

```text
Name: __________
Age:  __________
City: __________
```

The form stays the same.

The values change.

The structure stays the same.

# Chain of Thought (CoT)

**Chain of Thought** is a prompting technique that asks an AI to **solve a problem through a series of reasoning steps** before giving the final answer.

The main idea is:

> **Instead of asking only for the answer, guide the AI to reason through the problem step by step.**

---

## 1. What is Chain of Thought?

Imagine you ask:

> What is 25 × 16?

A simple prompt asks for the answer:

> **Answer:** 400

A Chain-of-Thought-style prompt asks the AI to work through the problem:

> Calculate 25 × 16 step by step.

The model may reason using intermediate steps and then produce:

> **Answer:** 400

The important idea is that **breaking a difficult problem into smaller reasoning steps can improve accuracy**.

---

## 2. Why is Chain of Thought useful?

CoT can be useful for problems that require multiple steps, such as:

- Mathematics
    
- Logic problems
    
- Debugging code
    
- Planning
    
- Data analysis
    
- Complex decision processes
    
- Multi-step reasoning

For example, suppose we ask:

> A product costs $80. It has a 20% discount. What is the final price?

There are two steps:

**Step 1:** Calculate the discount.

$80 × 20% = $16

**Step 2:** Subtract the discount.

$80 − $16 = $64

**Final answer: $64**

Breaking the problem into steps makes it easier to avoid mistakes.

---

## 3. A common Chain of Thought prompt

A traditional CoT prompt might say:

> **"Let's think step by step."**

For example:

> What is 15% of $200?
> **Let's think step by step.**

This encourages the model to perform multi-step reasoning.

However, for modern AI systems, you generally **don't need to ask the model to reveal its private chain-of-thought**. A better approach is to ask for a **concise explanation, key steps, or verification**.

---

## 4. Bad prompt example

Here is a weak prompt:

```text
Solve this problem:

A company has 120 employees.
25% work in IT.
40% of the IT employees work remotely.
How many IT employees work remotely?
```

### Why is it not ideal?

The question is understandable, but it doesn't explicitly tell the model how you want the solution presented.

For a learning or educational task, you may want:

- the calculation
    
- the intermediate result
    
- the final answer
    
- a clear explanation

---

## 5. Improved prompt example

A better prompt is:

```text
Solve this problem carefully.

A company has 120 employees.
25% work in IT.
40% of the IT employees work remotely.

Break the calculation into clear steps.
Show the important calculations.
Then give the final answer.
```

A possible response would be:

```text
Step 1: Calculate the number of IT employees.

120 × 25% = 30

Step 2: Calculate how many IT employees work remotely.

30 × 40% = 12

Final answer: 12 IT employees work remotely.
```

---

## 6. Why is the improved prompt better?

The improved prompt gives the AI a **clear process**.

### Bad prompt

```text
Solve this problem.
```

The model knows what to calculate, but the expected reasoning format is not specified.

### Improved prompt

```text
Break the calculation into clear steps.
Show the important calculations.
Then give the final answer.
```

Now the AI knows:

1. **Analyze the problem**
    
2. **Break it into smaller steps**
    
3. **Show the important calculations**
    
4. **Give the final answer**

This makes the response easier for a human to **check and understand**.

---

## 7. Chain of Thought vs. normal prompting

|Normal Prompt|Chain-of-Thought-style Prompt|
|---|---|
|"What is 25 × 16?"|"Calculate 25 × 16 carefully and show the key calculation."|
|Focuses on the answer|Focuses on the reasoning process|
|Good for simple questions|Useful for multi-step problems|
|Less explanation|More structured explanation|

---

## 8. An important point about modern AI

You may see prompts like:

```text
Think step by step and show all your reasoning.
```

In prompt engineering, it's better to distinguish between:

**Internal reasoning**
The model's private reasoning process.

**Explanation**
A concise, user-facing description of the important steps or justification.

For example, instead of asking:

> Show me your entire chain of thought.

A better prompt is:

```text
Solve the problem carefully.
Provide the key steps and calculations,
but keep the explanation concise.
```

This gives you the useful reasoning **without requiring the model to expose private internal reasoning**.

---

## 9. Simple formula to remember

You can think of Chain-of-Thought prompting as:

**Complex problem → smaller steps → solve each step → final answer**

For example:

```text
Problem
   ↓
Step 1
   ↓
Step 2
   ↓
Step 3
   ↓
Final answer
```

### Key takeaway

> **Chain of Thought is about encouraging structured, multi-step reasoning for complex tasks.**

When writing prompts, a practical pattern is:

```text
Solve the problem carefully.
Break it into clear steps.
Show the key calculations or reasoning.
Give the final answer.
```

This is often more useful than simply saying **"Give me the answer."**

# Tool Instructions

**Tool instructions** are rules that tell an AI **how and when to use external tools**.

A tool can let an AI do things that it cannot do with text alone, such as:

- Search the web
    
- Check the weather
    
- Query a database
    
- Read a file
    
- Call an API
    
- Send an email
    
- Calculate something
    
- Create a calendar event

The prompt needs to explain **what the tool does, when to use it, what information to give it, and how to use the result**.

---

## 1. What is a tool?

Think of an AI as a worker.

The AI can think and write, but sometimes it needs a **tool**.

For example:

```text
User: What is the weather in Sydney today?

AI: I need current weather information.
    ↓
Weather Tool
    ↓
Current weather data
    ↓
AI: Sydney is currently ...
```

The tool provides information or performs an action for the AI.

---

## 2. What are tool instructions?

**Tool instructions are instructions that control the AI's use of a tool.**

They usually answer questions like:

1. **When should I use the tool?**
    
2. **When should I NOT use the tool?**
    
3. **What information should I send to the tool?**
    
4. **How should I interpret the result?**
    
5. **What should I tell the user after using the tool?**

For example:

```text
Use the weather tool when the user asks about current
weather conditions.

Provide the user's city to the tool.

After receiving the result, summarize the weather
in simple English.
```

These are **tool instructions**.

---

## 3. Why are tool instructions important?

Without clear instructions, an AI may use tools incorrectly.

For example, imagine you have a database tool:

```text
search_customer(name)
```

A poor instruction might simply say:

```text
Use the customer tool.
```

The AI doesn't know:

- When should it use it?
    
- What information should it provide?
    
- What if the customer doesn't exist?
    
- Should it search by name or email?
    
- What should it do with the result?

Better instructions give the AI a clear process.

---

## 4. Tool instructions step by step

A useful way to write tool instructions is:

### Step 1 — Define the tool

Explain what the tool does.

```text
The customer tool searches the company
customer database.
```

### Step 2 — Define when to use it

```text
Use the customer tool when the user asks
for information about a customer.
```

### Step 3 — Define the required input

```text
Use the customer's email address when available.
```

### Step 4 — Define what to do with the result

```text
Use the returned customer information to answer
the user's question.
```

### Step 5 — Define error handling

```text
If no customer is found, tell the user that
the customer could not be found.
Do not invent customer information.
```

Now the AI has a much clearer workflow.

---

## 5. Bad prompt example

Suppose we have a weather tool.

### Bad prompt

```text
You have access to a weather tool.

Use the tool to answer weather questions.
```

This is too vague.

The AI doesn't know:

- When exactly to use the tool
    
- What information to send
    
- What to do if the user doesn't provide a location
    
- How to handle tool errors
    
- How to present the result

---

## 6. Improved prompt example

```text
You have access to a weather tool.

Tool instructions:

1. Use the weather tool whenever the user asks
   about current weather or today's weather.

2. Send the user's city or location to the weather tool.

3. If the user does not provide a location, ask them
   which city they mean before using the tool.

4. Use the information returned by the tool to answer
   the user's question.

5. Do not guess or invent current weather information.

6. If the tool fails, clearly tell the user that
   current weather information is unavailable.

7. Present the result in simple English and include
   the temperature and important weather conditions
   when available.
```

---

## 7. Why is the improved prompt better?

The improved prompt gives the AI a **clear procedure**.

|Bad prompt|Improved prompt|
|---|---|
|"Use the tool."|Explains **when** to use it|
|No input rules|Explains **what information** to send|
|No missing-data handling|Explains what to do if location is missing|
|No error handling|Explains what to do if the tool fails|
|Could guess information|Explicitly says **don't invent data**|
|No output instructions|Explains **how to present the result**|

The important idea is:

> **Don't just tell the AI that a tool exists. Tell it how to use the tool.**

---

## 8. A simple structure for tool instructions

You can use this template:

```text
Tool instructions:

1. PURPOSE
   Explain what the tool does.

2. WHEN TO USE
   Explain when the AI should call the tool.

3. INPUT
   Explain what information the AI should provide.

4. RESULT
   Explain how the AI should use the returned information.

5. ERRORS
   Explain what to do if the tool fails.

6. RESTRICTIONS
   Explain what the AI must not do.

7. OUTPUT
   Explain how the final answer should be presented.
```

For example:

```text
Tool instructions:

Purpose:
The database tool searches product information.

When to use:
Use it when the user asks about product price or stock.

Input:
Provide the product name.

Result:
Use the returned information to answer the question.

Errors:
If no product is found, tell the user.

Restrictions:
Never invent product prices or stock information.

Output:
Answer in simple English.
```

---

## 9. Tool instructions vs normal instructions

There is an important difference.

### Normal instruction

Tells the AI **how to answer**:

```text
Answer the user in simple English.
```

### Tool instruction

Tells the AI **how to interact with a tool**:

```text
Use the database tool when the user asks
for current product information.
```

They can work together:

```text
You are a helpful shopping assistant.

Answer in simple English.

Use the product database tool when the user
asks for current prices or stock.

Never invent product information.
```

---

## 10. Real-world example

Imagine you are building an AI customer-support system.

You have three tools:

```text
search_customer()
search_order()
create_support_ticket()
```

Your tool instructions might look like this:

```text
Tool instructions:

- Use search_customer when you need customer information.
- Use search_order when you need order information.
- Use create_support_ticket when the customer asks
  to report a problem that cannot be resolved immediately.

Before creating a support ticket, collect enough
information to describe the customer's problem.

Never invent customer or order information.

If a tool returns no result, tell the customer clearly.

After using a tool, explain the result in simple English.
```

This gives the AI a **tool-use strategy**.

---

## 11. The key idea

You can remember tool instructions with this simple formula:

**WHEN → INPUT → TOOL → RESULT → ACTION**

For example:

```text
WHEN:
The user asks for current weather.

INPUT:
City name.

TOOL:
Weather API.

RESULT:
Temperature and weather conditions.

ACTION:
Explain the current weather to the user.
```

So, in Prompt Engineering:

> **Tool instructions tell the AI when to use a tool, what information to provide, how to use the result, and what to do when something goes wrong.**

This becomes especially important when you build **AI agents, RAG systems, MCP applications, and tool-calling applications**, because the AI is no longer only generating text—it is also interacting with external systems.

# Guardrails

## 1. What are Guardrails?

**Guardrails** are rules or limits that tell an AI what it **should do, should not do, and how it should respond**.

Think of guardrails like the **safety rails on a road**:

- The road = what the AI is allowed to do.
    
- The guardrails = rules that keep the AI on the right path.
    
- The AI = the driver.

Without guardrails, an AI may produce answers that are unsafe, irrelevant, inaccurate, or outside the task.

---

## 2. Why do we need Guardrails?

Imagine you build a customer-support chatbot.

You want the chatbot to:

1. Answer questions about your products.
    
2. Be polite.
    
3. Only use information from your company documents.
    
4. Say "I don't know" when the information is unavailable.
    
5. Never invent prices or product information.

These are **guardrails**.

They help control the AI's behavior.

---

## 3. Common types of Guardrails

### A. Topic Guardrails

Tell the AI what topics it can discuss.

For example:

> Only answer questions about our company's products.

If someone asks:

> "Who will win the next election?"

The chatbot should not answer because this is outside its allowed topic.

---

### B. Information Guardrails

Tell the AI what information it can use.

For example:

> Only use information provided in the company knowledge base. If the answer cannot be found, say "I don't have enough information."

This helps reduce **hallucinations** — when an AI makes up information.

---

### C. Output Guardrails

Tell the AI how the answer should look.

For example:

> Return the answer as JSON with these fields:
> `answer`, `confidence`, and `source`.

This makes the output easier for a program to process.

---

### D. Safety Guardrails

Tell the AI what it must avoid.

For example:

> Do not provide instructions for illegal activities, harmful actions, or dangerous behavior.

---

### E. Role/Permission Guardrails

These control what an AI is allowed to do.

For example, imagine an AI assistant connected to a company's database.

You might specify:

> You can read customer information, but you cannot delete or modify customer records.

This is especially important when an AI has access to **tools or external systems**.

---

## 4. Bad Prompt Example

Here is a weak prompt:

```text
You are a customer service chatbot.

Answer the customer's questions.
```

The problem is that this gives the AI very few boundaries.

For example, the AI might:

- Make up product information.
    
- Answer questions unrelated to the company.
    
- Give information that isn't in the company's database.
    
- Give very long answers.
    
- Guess when it doesn't know the answer.

---

## 5. Improved Prompt with Guardrails

Here is a better prompt:

```text
You are a customer service assistant for ABC Electronics.

Your job is to answer questions about ABC Electronics products and services.

Follow these rules:

1. Only answer questions related to ABC Electronics.
2. Use only the information provided in the company knowledge base.
3. Never invent product information, prices, features, or policies.
4. If the information is not available, say:
   "I'm sorry, but I don't have enough information to answer that."
5. Do not guess or make assumptions.
6. Keep answers short and easy to understand.
7. If a customer asks about something outside your role, politely explain that you can only help with ABC Electronics products and services.
```

---

## 6. Why is the improved prompt better?

The improved prompt gives the AI **clear boundaries**.

|Rule|Purpose|
|---|---|
|Only answer company questions|Controls the topic|
|Use the knowledge base|Controls the information source|
|Never invent information|Reduces hallucinations|
|Say "I don't have enough information"|Gives the AI a safe fallback|
|Don't guess|Improves reliability|
|Keep answers short|Controls the output|
|Stay within the role|Prevents unrelated responses|

The important idea is:

> **A good prompt doesn't only tell the AI what to do. It also tells the AI what NOT to do.**

---

## 7. Guardrails + Prompt Engineering

You can think about a good prompt like this:

```text
ROLE
  ↓
What is the AI?

TASK
  ↓
What should it do?

RULES
  ↓
What must it follow?

GUARDRAILS
  ↓
What must it NOT do?

FALLBACK
  ↓
What should it do when it doesn't know?

OUTPUT FORMAT
  ↓
How should it respond?
```

For example:

```text
Role:
You are a customer support assistant.

Task:
Answer customer questions.

Guardrails:
- Only discuss our products.
- Don't invent information.
- Don't guess.
- Don't provide information outside the knowledge base.

Fallback:
If you don't know the answer, say "I don't have enough information."

Output:
Give a short, clear answer.
```

## 8. The key idea

**Guardrails = boundaries for AI behavior.**

They help make AI systems:

- **Safer**
    
- **More reliable**
    
- **More predictable**
    
- **More focused**
    
- **Less likely to hallucinate**
    
- **Easier to integrate into applications**

One important point: **prompt guardrails are only one layer of protection**. In a real AI application, important guardrails can also be implemented in code—for example, validating tool arguments, checking permissions, filtering inputs/outputs, and requiring confirmation before sensitive actions.
