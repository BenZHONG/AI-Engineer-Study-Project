# Node

```python
from llama_index.core import Document
from llama_index.core.node_parser import SentenceSplitter

document = Document(
    text="""
    Employees receive 20 days of annual leave.
    Leave must be approved by a manager.
    Employees should submit requests at least two weeks in advance.
    Sick leave is managed separately from annual leave.
    """
)

parser = SentenceSplitter(
    chunk_size=100,
    chunk_overlap=20,
)

nodes = parser.get_nodes_from_documents([document])

for node in nodes:
    print(node.get_content())
```

# Index

```python
from llama_index.core import Document, VectorStoreIndex

documents = [
    Document(text="Employees receive 20 days annual leave."),
    Document(text="Employees must use MFA for company systems."),
]

index = VectorStoreIndex.from_documents(documents)
```

```python
query_engine = index.as_query_engine()

response = query_engine.query("How many days of annual leave do employee recive?")

print(response)
```
